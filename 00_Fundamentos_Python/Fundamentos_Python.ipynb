# 1. Variables

## 1.1 Qué es una variable

Una variable es un **nombre que hace referencia a un valor**. Una variable crea la primera vez que se le asigna un valor mediante el operador `=`. 

A diferencia de lenguajes como **Java** En Python no se declara el tipo. El tipo de datos de una variable lo determina el valor asignado. 

````python
a = 14                  # Tipo entero (int)
b = 3.5                 # Tipo decimal (float)
nombre = "Juan"         # Tipo cadena de texto (str)
es_estudiante = True    # Tipo booleano (bool)

print(a, b, nombre, es_estudiante)
````

## 1.2 Tipos de las variables

En Python el tipo pertenece al **valor**, no a la variable. Una misma variable puede guardar después un valor de otro tipo (se dice que Python tiene **tipado dinámico**). En todo caso, no conviene que una variable cambie de tipo: el código se vuelve difícil de seguir.

````python
a = 14
print(a, type(a))
a = "Hola"
print(a, type(a))
````

## 1.3 Nombres de variables

**Reglas** (si no se cumplen, da error):

* Pueden contener letras, números y el guion bajo `_`.
* **No pueden empezar por un número**: `2precio` no es válido.
* No pueden tener espacios ni guiones: `precio-final` se interpreta como una resta.
* No pueden ser **palabras reservadas** del lenguaje (`if`, `for`, `True`, `class`, `import`…).
* Distinguen **mayúsculas y minúsculas**: `edad`, `Edad` y `EDAD` son tres variables distintas.

**Convenciones** (no son obligatorias, pero todo el mundo las sigue; guía de estilo PEP 8):

* Nombres en minúscula con palabras separadas por guion bajo (*snake_case*): `precio_final`, `num_clientes`.
* Nombres que digan qué contienen: `ventas_2025` mejor que `v2`.
* Las constantes (valores que no cambian), en mayúsculas: `IVA = 0.21`.

Las palabras reservadas se pueden consultar así:

````python
import keyword
print(keyword.kwlist)
````

## 1.4 Asignación múltiple e intercambio
Python es un lenguaje con una sintaxis muy expresiva. Python permite asignar valores a varias variables a la vez, y también intercambiar valores entre variables sin necesidad de una variable temporal. 

````python
x, y, z = 1, 2, 3        # varias variables en una línea
print(x, y, z)

x, y = y, x              # intercambiar dos valores sin variable auxiliar
print(x, y)

a = b = 0                # el mismo valor a varias variables
print(a, b)
````

## 1.5 Comentarios

Todo lo que va detrás del carácter `#` es un **comentario**: Python lo ignora. Los comentarios sirven para explicar el **porqué** del código, no para repetir lo que ya se lee.

````python
IVA = 0.21                   # tipo general en España
precio = 100
precio_final = precio * (1 + IVA)
print(precio_final)
````

# 2. El tipo de dato: type() e isinstance()

## 2.1 `type()`

La **función** `type()` devuelve el tipo de un valor:

````python
print(type(14))        # <class 'int'>
print(type(3.5))       # <class 'float'>
print(type("Hola"))    # <class 'str'>
print(type(True))      # <class 'bool'>
print(type(None))      # <class 'NoneType'>
````

## 2.2 `isinstance()`

`isinstance(valor, tipo)` devuelve `True` si el valor es de ese tipo. Es la forma recomendada de **comprobar** un tipo dentro de un programa. Admite una **tupla de tipos** para comprobar varios a la vez.

````python
edad = 34
print(isinstance(edad, int))              # True
print(isinstance(edad, str))              # False
print(isinstance(3.5, (int, float)))      # True: es alguno de los dos
````

> **Ojo.** En Python, `bool` es un subtipo de `int`, así que `isinstance(True, int)` devuelve `True`. Si hay que distinguirlos, se comprueba primero si es `bool`.

````python
print(isinstance(True, int))     # True
print(type(True) == int)         # False: el tipo exacto es bool
print(type(True) == bool)         # True: el tipo exacto es bool
````

# 3. Números enteros: int

El tipo de los enteros es `int`. Python tiene **un solo tipo de entero**, y su tamaño solo está limitado por la memoria: no se desborda como en otros lenguajes (ni como en NumPy, que sí tiene enteros de tamaño fijo).

````python
grande = 123456789012345678901234567890
print(grande, type(grande))
print(2 ** 1000)                # 2 elevado a 1000, sin problema
````

## 3.1 Escribir números legibles

El guion bajo `_` se puede usar como **separador de miles** al escribir números en el código. Python ignora ese símbolo para proporcionar una mejor legibilidad.

````python
poblacion = 47_000_000
presupuesto = 2_500_000
print(poblacion, presupuesto)
````

## 3.2 Otras bases

Los enteros se pueden escribir en hexadecimal (`0x`), octal (`0o`) y binario (`0b`). Las funciones `hex()`, `oct()` y `bin()` hacen la conversión inversa.

````python
a = 0x1A        # hexadecimal: 26
b = 0o12        # octal: 10
c = 0b1011      # binario: 11
print(a, b, c)
print(hex(26), oct(10), bin(11))
````

# 4. Números decimales: float

Los decimales son de tipo `float` y se escriben **con punto** como separador decimal. Python tiene un solo tipo decimal, de doble precisión (64 bits), equivalente al `double` de otros lenguajes. Tiene unos **15-16 dígitos significativos**.

````python
pi_aprox = 3.14159
precio = 2.05
print(type(pi_aprox), type(precio))
print(type(7 / 2))      # la división / siempre da float
````

## 4.1 Notación científica

`e` significa «por diez elevado a»: útil para números muy grandes o muy pequeños.

````python
a = 1.5e3       # 1,5 · 10³ = 1500.0
b = 2.5e-4      # 2,5 · 10⁻⁴ = 0.00025
print(a, b)
````

## 4.2 La precisión de los decimales

Los decimales se guardan en binario en notación de punto flotante, y muchos números sencillos (como 0,1) no tienen una representación exacta. Esto produce pequeños errores de redondeo. **Por eso no se comparan decimales con `==`**: se usa `math.isclose()`.

````python
import math # Necesario para usar math.isclose()

print(0.1 + 0.2)                        # 0.30000000000000004
print(0.1 + 0.2 == 0.3)                 # False
print(math.isclose(0.1 + 0.2, 0.3))     # True
````

## 4.3 Redondear: `round()`

`round(número, decimales)` redondea. Dos detalles que sorprenden:

* Cuando el número está **justo en la mitad**, `round` redondea al **par más cercano** (*redondeo bancario*): `round(2.5)` da 2 y `round(3.5)` da 4.
* Por la precisión binaria de los decimales, `round(2.675, 2)` da `2.67`: el 2,675 se guarda en realidad como 2,67499999…

Para **mostrar** un número con un número fijo de decimales es mejor usar el formato de las cadenas f (apartado 7.6), que no modifica el valor.

````python
print(round(3.14159, 2))     # 3.14
print(round(2.5), round(3.5))  # 2 4
print(round(2.675, 2))        # 2.67
````

# 5. Valores booleanos: bool

El tipo `bool` tiene solo dos valores: `True` (verdadero) y `False` (falso), escritos **con la primera letra en mayúscula**. Los devuelven las comparaciones y las expresiones lógicas.

````python
a = True
b = False
c = 5 > 3        # True
d = 2 == 4       # False
print(a, b, c, d)
````

## 5.1 Valores que equivalen a verdadero o falso

Cualquier valor se puede usar donde se espera un booleano (por ejemplo, en un `if`). Python lo interpreta así:

| Se consideran **falsos** | Se consideran **verdaderos** |
|---|---|
| `False` | `True` |
| `None` | Todo lo demás |
| `0`, `0.0` | Cualquier número distinto de cero (también los negativos) |
| `""` (cadena vacía) | Cualquier cadena no vacía (también `" "` incluida la cadena `"False"`) |
| `[]`, `()`, `{}` (colecciones vacías) | Cualquier colección con algún elemento |

````python
print(bool(12), bool(-3), bool(0))        # True True False
print(bool("Hola"), bool(""))             # True False
print(bool(None))                         # False
print(bool([1, 2]), bool([]))             # True False
````

## 5.2 Los booleanos son números

`True` vale 1 y `False` vale 0. Por eso **sumar booleanos cuenta los verdaderos**, algo muy útil en análisis de datos (y que NumPy y pandas aprovechan).

````python
print(True + True + False)       # 2
respuestas = [True, False, True, True]
print("Respuestas afirmativas:", sum(respuestas))
````

# 6. El valor None

`None` representa **la ausencia de valor**: «no hay nada», «no consta». Es el único valor de su tipo (`NoneType`). Lo devuelven, por ejemplo, las funciones que no tienen `return`.

Para comprobar si algo es `None` se usa **`is`**, no `==`.

````python
telefono = None
print(telefono, type(telefono))
print(telefono is None)         # True: la forma correcta de comprobarlo

resultado = print("Hola")       # print no devuelve nada...
print(resultado)                # ...así que resultado es None
````

> En análisis de datos, la ausencia de un dato se representa normalmente con `NaN` (NumPy) o `pd.NA` (pandas), no con `None`. Se verá en esos apuntes.

# 7. Cadenas de caracteres: str

## 7.1 Escribir cadenas

Las cadenas de texto (`str`) van entre comillas simples (`'...'`) o dobles (`"..."`); es indiferente, pero hay que cerrar con la misma que se abre. Las **comillas triples** (`'''...'''` o `"""..."""`) permiten escribir cadenas de **varias líneas**.

````python
texto1 = 'Hola, mundo!'
texto2 = "Python es genial."
texto3 = "Dijo: 'hoy no'"          # unas comillas dentro de las otras
texto4 = '''Esta es una cadena
de varias líneas.'''
print(texto1)
print(texto2)
print(texto3)
print(texto4)
````

## 7.2 Caracteres especiales

Dentro de una cadena, la barra invertida `\` introduce **caracteres especiales**:

| Secuencia | Significado |
|---|---|
| `\n` | Salto de línea |
| `\t` | Tabulador |
| `\\` | Una barra invertida |
| `\'` y `\"` | Comillas dentro de una cadena delimitada por las mismas comillas |

Una cadena precedida de **`r`** (*raw*, «en bruto») no interpreta las barras. Es útil para escribir **rutas de archivos**.

````python
print("Nombre:\tAna\nEdad:\t34")
print("C:\\Users\\alumno")          # barras dobles
print(r"C:\Users\alumno\nuevo")       # cadena en bruto: \n no es un salto de línea
````

## 7.3 Longitud, índices y rebanadas

Una cadena es una **secuencia** de caracteres:

* `len(cadena)` da su longitud.
* Cada carácter tiene un **índice** que empieza en 0; los índices negativos cuentan desde el final.
* Las **rebanadas** `cadena[inicio:fin]` permiten extraer un trozo del string, desde el `inicio`  hasta el `fin` (el `fin` no se incluye).
* También se pueden usar **pasos**: `cadena[inicio:fin:paso]` toma un carácter cada `paso`. Si `paso`vale `2`, toma un carácter cada dos.
* Un paso negativo recorre la cadena al revés.
* Si uno de los tres números (`inicio`, `fin`, `paso`) se omite, Python asume el valor por defecto: `inicio`=0, `fin`=longitud de la cadena, `paso`=1.

````python
provincia = "Valladolid"
print(len(provincia))        # 10
print(provincia[0])          # V
print(provincia[-1])         # d
print(provincia[0:4])        # Vall
print(provincia[4:])         # adolid
print(provincia[1:6:2])      # ald
print(provincia[::-1])       # al revés
````

## 7.4 Las cadenas no se pueden modificar

Las cadenas son **inmutables**: no se puede cambiar un carácter. Los métodos que «modifican» una cadena en realidad **devuelven una cadena nueva**; si no se guarda, el cambio se pierde.

````python
nombre = "ana"
try:
    nombre[0] = "A"  # No podemos cambiar un carácter de una cadena: son inmutables
except TypeError as e:
    print("TypeError:", e)

nombre.upper()          # devuelve "ANA", pero no se guarda
print(nombre)           # sigue siendo "ana"
nombre = nombre.upper() # así sí
print(nombre)
````

## 7.5 Métodos útiles de las cadenas

Son la base de la **limpieza de textos** en análisis de datos (en pandas se usan casi igual, con `.str`).

| Método | Qué hace |
|---|---|
| `upper()`, `lower()`, `title()` | Mayúsculas, minúsculas, primera letra de cada palabra en mayúscula |
| `strip()` | Quita los espacios del principio y del final (`lstrip`, `rstrip`: quitan solo el espacio al principio o al final respectivamente) |
| `replace(a, b)` | Sustituye `a` por `b` |
| `split(sep)` | Divide el texto en una lista usando el separador `sep`, donde cada elemento es un trozo |
| `sep.join(lista)` | Une una lista de cadenas con el separador |
| `startswith(x)`, `endswith(x)` | Devuelve True si el texto empieza o termina por `x`, respectivamente |
| `find(x)` | Posición de la primera aparición de `x` (−1 si no está) |
| `count(x)` | Indica cuántas veces aparece el texto `x` en el string |
| `isdigit()` | `True` si todos los caracteres son dígitos, `False` en caso contrario |

````python
dato = "  león  "
print("[" + dato.strip() + "]")
print(dato.strip().title())                    # León
print("1.234,56".replace(".", "").replace(",", "."))   # 1234.56
print("Ana;34;León".split(";"))                # ['Ana', '34', 'León']
print(" | ".join(["León", "Soria", "Ávila"]))
print("ventas_2025.csv".endswith(".csv"))      # True
print("Valladolid".count("l"))                 # 3
print("2025".isdigit(), "20,5".isdigit())      # True False
````

El operador **`in`** comprueba si un texto contiene a otro:

````python
print("León" in "Calle de León, 3")    # True
print("leon" in "Calle de León, 3")    # False: distingue mayúsculas y tildes
````

## 7.6 Cadenas f (*f-strings*)

Es la forma recomendada de **construir textos con valores dentro**. Para ello, se indica una `f` delante de la cadena. Las expresiones procedentes del lenguaje Python se indican entre llaves `{}`. 

Dentro de esas llaves podemos indicar dos puntos (:) detrás de los cuales se indica el **formato** con el que queremos mostrar los valores.

````python
nombre = "Ana"
edad = 34
print(f"{nombre} tiene {edad} años")
print(f"Dentro de diez años tendrá {edad + 10}")     # admite expresiones
````

| Formato | Ejemplo | Resultado |
|---|---|---|
| `{x:.2f}` | 2 decimales | `3.14` |
| `{x:,.2f}` | separador de miles y 2 decimales | `1,234,567.89` |
| `{x:.1%}` | porcentaje con 1 decimal | `25.6%` |
| `{x:>8}` | alineado a la derecha en 8 posiciones | `      42` |
| `{x:<8}` | alineado a la izquierda | `42      ` |
| `{x:08d}` | entero con ceros a la izquierda | `00000042` |

````python
importe = 1234567.891
print(f"{importe:.2f}")
print(f"{importe:,.2f}")
print(f"{0.256:.1%}")
print(f"[{42:>8}]  [{42:<8}]  [{42:08d}]")
print(f"{'León':<12}{1500:>10,}")       # columnas alineadas
````

> **Formato español.** Las cadenas f usan el formato inglés (coma para miles, punto para decimales). Para mostrar `1.234.567,89` hay que intercambiar los símbolos mediante el método `replace()`.

````python
texto = f"{importe:,.2f}"
print(texto.replace(",", "X").replace(".", ",").replace("X", "."))
````

# 8. Conversión de tipos

Para convertir un valor a otro tipo se usa la función con el nombre del tipo: `int()`, `float()`, `bool()` y `str()`. Si la conversión no es posible, se produce un **`ValueError`**.

> En las celdas siguientes, los errores se capturan con `try`/`except` para que el cuaderno se pueda ejecutar entero. Sin eso, el error detendría la ejecución.

## 8.1 Conversión a enteros: `int()`

````python
print(int(8.9))         # 8: trunca, no redondea
print(int(-8.9))        # -8: trunca hacia el cero
print(int("123"))       # 123
print(int(" 42 "))      # 42: admite espacios alrededor
print(int(True))        # 1
print(int(False))       # 0
````

````python
for texto in ["Hola", "3.5", "1.000"]:
    try:
        int(texto)
    except ValueError as e:
        print("ValueError:", e)
````

> `int("3.5")` falla: un texto con decimales hay que convertirlo antes a `float`: `int(float("3.5"))`. Y `int("1.000")` también falla: el punto como separador de miles no es válido.

## 8.2 Conversión a decimales: `float()`

````python
print(float(8))          # 8.0
print(float("3.14"))     # 3.14
print(float("1e3"))      # 1000.0
print(float(True))       # 1.0
print(float(False))      # 0.0
````

````python
for texto in ["Hola", "3,14"]:
    try:
        float(texto)
    except ValueError as e:
        print("ValueError:", e)

print(float("3,14".replace(",", ".")))     # la solución para la coma decimal
````

> **Importante en análisis de datos.** Los ficheros generados en España suelen usar la **coma decimal** (`3,14`). Python solo entiende el **punto**. Es uno de los problemas más frecuentes al leer datos, y pandas tiene un parámetro específico para resolverlo.

## 8.3 Conversión a booleanos: `bool()`

Sigue la tabla del apartado 5.1:

````python
print(bool(8))         # True
print(bool(0))         # False
print(bool(3.14))      # True
print(bool(0.0))       # False
print(bool("Hola"))    # True
print(bool(""))        # False
print(bool("False"))   # True: es una cadena no vacía
print(bool(" "))       # True: un espacio no es una cadena vacía
````

> **Ojo.** `bool("False")` es `True`. Para convertir un texto como `"sí"`/`"no"` o `"True"`/`"False"` en booleano hay que **compararlo**: `texto == "True"`.

## 8.4 Conversión a cadenas: `str()`

````python
print(str(123))        # 123
print(str(3.14))       # 3.14
print(str(True))       # True
print(repr(str(123)))  # '123': repr muestra las comillas, para ver que es texto
````

`print` muestra el texto sin comillas, aunque el valor sea una cadena. Para ver la diferencia entre `123` y `"123"`, se usa `type()` o `repr()`.

# 9. Operadores

## 9.1 Operadores aritméticos

| Operador | Operación | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `14 + 5` | `19` |
| `-` | Resta | `14 - 5` | `9` |
| `*` | Multiplicación | `14 * 5` | `70` |
| `/` | División (siempre da `float`) | `14 / 5` | `2.8` |
| `//` | División entera | `14 // 5` | `2` |
| `%` | Resto (módulo) | `14 % 5` | `4` |
| `**` | Potencia | `2 ** 3` | `8` |

````python
print(14 / 5)     # 2.8
print(14 // 5)    # 2
print(14 % 5)     # 4
print(2 ** 3)     # 8
print(10 / 2)     # 5.0: / siempre devuelve float
print(9 ** 0.5)   # 3.0: raíz cuadrada como potencia
````

> **Ojo con los negativos.** `//` redondea **hacia abajo** (no hacia el cero), y `%` toma el signo del divisor: `-7 // 2` da `-4` y `-7 % 2` da `1`.

````python
print(-7 // 2, -7 % 2)     # -4 1

try:
    print(10 / 0)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
````

**El operador `%` sirve para saber si un número es divisible por otro**: `n % 2 == 0` es cierto para los pares.

## 9.2 Prioridad de los operadores

De mayor a menor prioridad:

| Prioridad | Operadores |
|---|---|
| 1 | `( )` paréntesis |
| 2 | `**` |
| 3 | `-x` (signo negativo) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `in`, `is` |
| 7 | `not` |
| 8 | `and` |
| 9 | `or` |

Ante la duda, **paréntesis**: hacen el código más claro y no cuestan nada.

````python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(-2 ** 2)         # -4: la potencia va antes que el signo
print((-2) ** 2)       # 4
````

## 9.3 Operadores de comparación

Devuelven `True` o `False`.

| Operador | Significado |
|---|---|
| `==` | Igual |
| `!=` | Distinto |
| `>`, `<` | Mayor, menor |
| `>=`, `<=` | Mayor o igual, menor o igual |

````python
print(5 == 5)     # True
print(5 != 3)     # True
print(7 > 10)     # False
print(4 < 8)      # True
print(6 >= 6)     # True
print(3 <= 2)     # False
````

Las comparaciones se pueden **encadenar**, como en matemáticas:

````python
edad = 34
print(18 <= edad < 65)      # True: equivale a 18 <= edad and edad < 65
````

Las cadenas se comparan **por orden alfabético** (según el código de cada carácter: las mayúsculas van antes que las minúsculas y las letras con tilde, después de todas).

````python
print("Ana" < "Luis")      # True
print("ana" < "Luis")      # False: las mayúsculas van antes
print("Ávila" < "Burgos")  # False: la Á va después de todas las letras sin tilde
````

> **`==` frente a `is`.** `==` compara **valores**; `is` comprueba si son **el mismo objeto**. Se usa `is` solo para `None` (`x is None`).

## 9.4 Operadores lógicos

| Operador | Resultado |
|---|---|
| `a and b` | `True` si **los dos** son verdaderos |
| `a or b` | `True` si **alguno** es verdadero |
| `not a` | Lo contrario de `a` |

````python
print(True and False)          # False
print(True or False)           # True
print(not True)                # False
print(17 > 12 and 5 < 10)      # True
print(3 == 3 or 4 != 4)        # True
````

**Evaluación en cortocircuito.** Python deja de evaluar en cuanto conoce el resultado: en `a and b`, si `a` es falso, `b` ni se mira. Permite escribir condiciones seguras:

````python
divisor = 0
print(divisor != 0 and 10 / divisor > 1)   # False, sin error: la división no se llega a hacer
````

## 9.5 Operadores de asignación

| Operador | Equivale a |
|---|---|
| `x = 5` | Asignación simple |
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |
| `x //= 5` | `x = x // 5` |
| `x %= 5` | `x = x % 5` |
| `x **= 5` | `x = x ** 5` |

````python
x = 29
x += 5       # x = x + 5
print(x)     # 34
x *= 2       # x = x * 2
print(x)     # 68
x //= 3      # x = x // 3
print(x)     # 22
````

EN Python no hay operadores de incremento/decremento (`++`, `--`): se usan `+= 1` y `-= 1`.

## 9.6 Operadores con cadenas

**Concatenación (`+`).** Une cadenas. Todos los operandos tienen que ser cadenas: un número hay que convertirlo con `str()` (o, mejor, usar una cadena f).

````python
print("Hola, " + "mundo!")
nombre = "Ana"
print("Hola, " + nombre + "!")

numero = 19
try:
    print("El número es: " + numero)
except TypeError as e:
    print("TypeError:", e)
print("El número es: " + str(numero))
print(f"El número es: {numero}")        # más cómodo
````

**Repetición (`*`).** Repite una cadena un número de veces.

````python
print("Hola" * 3)
print("-" * 40)
````

# 11. Librerías

Una **librería** (o **módulo**) es un conjunto de funciones y valores ya programados que se pueden usar en un programa tras **importarlos**.

* La **biblioteca estándar** viene con Python: `math`, `random`, `statistics`, `datetime`, `csv`… No hay que instalar nada.
* Las **librerías externas** (NumPy, pandas, Matplotlib…) hay que **instalarlas** en el entorno del proyecto.

## 11.1 Formas de importar

````python
import math                        # se importa el módulo entero
print(math.sqrt(16))               # se usa con el prefijo del módulo
print(math.pi)
````

````python
from math import sqrt, pi          # se importan solo algunos elementos
print(sqrt(9), pi)                 # se usan sin prefijo
````

````python
import math as m                   # se importa con un alias (nombre corto)
print(m.sqrt(9))
````

> La forma `from math import *` importa todo sin prefijo. **No se recomienda**: no se sabe de dónde viene cada nombre y puede chocar con nombres de otras librerías.

## 11.2 La librería `math`

````python
import math

print(math.sqrt(2))          # raíz cuadrada
print(math.floor(3.7))       # 3: redondeo hacia abajo
print(math.ceil(3.2))        # 4: redondeo hacia arriba
print(math.log(math.e))      # 1.0: logaritmo natural
print(math.log10(1000))      # 3.0: logaritmo en base 10
print(math.fabs(-5))         # 5.0: valor absoluto (también existe abs())
````

## 11.3 La librería `random`

Genera números **pseudoaleatorios**. Con `random.seed(n)` se fija la **semilla**: con la misma semilla se obtienen siempre los mismos números, y el resultado es **reproducible**.

| Función | Qué devuelve |
|---|---|
| `random.random()` | Un decimal entre 0 y 1 (el 1 no se incluye) |
| `random.randint(a, b)` | Un entero entre `a` y `b`, **ambos incluidos** |
| `random.uniform(a, b)` | Un decimal entre `a` y `b` |
| `random.choice(secuencia)` | Un elemento al azar |
| `random.sample(secuencia, k)` | `k` elementos distintos al azar (muestra sin reemplazo) |
| `random.shuffle(lista)` | Desordena la lista en el sitio |

````python
import random

random.seed(42)
print("Decimal entre 0 y 1:", random.random())
print("Entero entre 1 y 10:", random.randint(1, 10))
print("Una provincia:", random.choice(["León", "Soria", "Ávila", "Burgos"]))
print("Muestra de 3 clientes:", random.sample(range(1, 101), 3))
````

> En análisis de datos, para generar muchos números aleatorios se usa el generador de NumPy (`np.random.default_rng`), que se explica en sus apuntes.

# 12. Errores frecuentes

| Lo que ves | Qué pasa | Solución |
|---|---|---|
| `NameError: name 'x' is not defined` | La variable no existe: no se ha ejecutado la celda que la crea, o está mal escrita (mayúsculas) | Ejecuta las celdas en orden; revisa el nombre |
| `SyntaxError: invalid syntax` | Algo mal escrito: comillas o paréntesis sin cerrar, `=` en lugar de `==`… | Mira la línea que señala el mensaje |
| `TypeError: can only concatenate str (not "int") to str` | Se suma un texto y un número | `str(numero)` o una cadena f |
| `ValueError: invalid literal for int() with base 10` | El texto no se puede convertir a número | Revisa el texto: decimales, espacios internos, letras |
| `ValueError: could not convert string to float: '3,14'` | Coma decimal | `.replace(",", ".")` antes de convertir |
| `ZeroDivisionError: division by zero` | División entre cero | Comprueba el divisor antes |
| `0.1 + 0.2 == 0.3` es `False` | Precisión de los decimales | `math.isclose()` |
| `bool("False")` es `True` | Toda cadena no vacía es verdadera | Compara: `texto == "True"` |
| `"21" * 2` da `"2121"` | `input` devuelve texto | `int(input(...))` |
| `TypeError: 'str' object does not support item assignment` | Las cadenas no se pueden modificar | Crea una cadena nueva |
| `ModuleNotFoundError: No module named '…'` | La librería no está instalada en el entorno | `uv add …` y arrancar con `uv run` |
