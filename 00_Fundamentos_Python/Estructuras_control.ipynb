# 1. Los bloques y la sangría

En Python, las instrucciones que forman parte de un condicional o de un bucle se agrupan en un **bloque**. El bloque empieza después de los **dos puntos** (`:`) y las instrucciones del bloque se marcan con una **sangría**: todas sus líneas van desplazadas a la derecha **el mismo número de espacios** (por convención, **4**). El bloque termina en la primera línea que deja de usar la sangría.

```python
if condición:
    instrucción 1      # dentro del bloque
    instrucción 2      # dentro del bloque
instrucción 3          # fuera: se ejecuta siempre
```

La sangría **no es decorativa**: cambia el significado del programa. Una sangría incoherente produce un error.

````python
codigo = "if True:\n    print('uno')\n      print('dos')"     # la segunda línea tiene 6 espacios
try:
    exec(codigo) # ejecuta el código de la cadena 
except IndentationError as e:
    print("IndentationError:", e)
````

> Jupyter y VS Code ponen la sangría solos al pulsar Intro después de `:`. Para quitar un nivel, Retroceso; para sangrar varias líneas a la vez, se seleccionan y se pulsa Tab (o Mayús + Tab para quitarla).

`pass` es una instrucción que **no hace nada**. Sirve para dejar un bloque vacío, por ejemplo, mientras se desarrolla el resto del programa.

````python
nota = 7
if nota >= 5:
    pass          # pendiente de escribir
print("El programa sigue")
````

# 2. Condicionales

## 2.1 `if` simple

Ejecuta un bloque **solo si** se cumple una condición.

```python
if condición:
    # bloque que se ejecuta si la condición es verdadera
```

````python
numero = 7
if numero > 0:
    print("El número es positivo")
print("Fin")        # fuera del if: se ejecuta siempre
````

## 2.2 `if-else`

Ejecuta un bloque si se cumple la condición y **otro** si no se cumple. Siempre se ejecuta uno de los dos, y solo uno.

```python
if condición:
    # bloque si la condición es verdadera
else:
    # bloque si la condición es falsa
```

````python
numero = -3
if numero > 0:
    print("El número es positivo")
else:
    print("El número es cero o negativo")
````

## 2.3 `if-elif-else` (condiciones múltiples)

Evalúa varias condiciones **en orden**. `elif` es la contracción de *else if*: «si no se cumplió lo anterior, prueba esto». **Se ejecuta solo el primer bloque cuya condición se cumpla**; las demás ya no se comprueban. El `else` final, que es opcional, recoge los casos en los que no se cumple ninguna.

```python
if condición1:
    # si condición1 es verdadera
elif condición2:
    # si condición1 es falsa y condición2 es verdadera
# … más elif
else:
    # si ninguna es verdadera
```

````python
nota = 7.5
if nota >= 9:
    print("Sobresaliente")
elif nota >= 7:
    print("Notable")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
````

> **El orden importa.** Como solo se ejecuta el primer bloque que se cumple, las condiciones tienen que ir de la más restrictiva a la menos. Si la primera fuera `nota >= 5`, un 9,5 saldría como «Aprobado».

````python
nota = 9.5
if nota >= 5:          # orden incorrecto
    print("Aprobado")
elif nota >= 9:
    print("Sobresaliente")   # nunca se llega aquí
````

## 2.4 `if` anidado

Un `if` **dentro** de otro. Sirve cuando la segunda pregunta solo tiene sentido si la primera se cumple.

````python
edad = 20
tiene_carne = False

if edad >= 18:
    if tiene_carne:
        print("Puede conducir")
    else:
        print("Es mayor de edad, pero no tiene carné")
else:
    print("Es menor de edad")
````

> Muchos `if` anidados se pueden escribir mejor con **condiciones compuestas** (`and`, `or`, `not`) o con `elif`. Más de dos o tres niveles de anidamiento hacen el código difícil de leer.

## 2.5 Condiciones compuestas

Las condiciones se combinan con `and`, `or` y `not`, y admiten comparaciones encadenadas:

````python
edad = 34
provincia = "León"

if 18 <= edad < 65 and provincia in ("León", "Zamora", "Salamanca"):
    print("Cliente adulto de la zona oeste")

if not (edad < 18):
    print("Mayor de edad")
````

Recuerda que cualquier valor se puede usar como condición: `0`, `""`, `None` y las colecciones vacías se consideran **falsos** (ver los apuntes de conceptos básicos):

````python
nombre = ""
if nombre:
    print(f"Hola, {nombre}")
else:
    print("No has escrito ningún nombre")
````

## 2.6 Expresión condicional (operador ternario)

Se trata de una versión abreviada de una estructura `if-else` que permite asignar un valor a una variable según una condición. Se realiza en una sola línea y tiene la forma:

```python
valor = valor_si_verdadero if condición else valor_si_falso
```

````python
nota = 4
resultado = "Aprobado" if nota >= 5 else "Suspenso"
print(resultado)
````

## 2.7 `match` (Python 3.10 o posterior)

Compara un valor con varios **patrones**. Es útil cuando hay muchos casos que dependen del valor exacto de una variable. `_` es el caso por defecto, y `|` agrupa varios valores.

````python
dia = "sábado"
match dia:
    case "sábado" | "domingo":
        print("Fin de semana")
    case "lunes":
        print("Empieza la semana")
    case _:
        print("Día laborable")
````

# 3. Bucle while

Repite un bloque **mientras** se cumpla una condición. La condición se comprueba **antes** de cada vuelta: si es falsa desde el principio, el bloque no se ejecuta ninguna vez.

```python
while condición:
    # bloque que se repite
```

````python
i = 1
while i <= 10:        # imprime los números del 1 al 10
    print(i, end=" ")
    i += 1            # sin esta línea, el bucle no terminaría nunca
````

> **Bucle infinito.** Si dentro del bucle nada cambia la condición, el bucle no termina. En Jupyter la celda se queda en `[*]`: se detiene con *Kernel → Interrupt Kernel* (o **I, I** en modo comando).

`while` se usa cuando **no se sabe de antemano cuántas vueltas** hay que dar en el bucle. Por ejemplo: ¿cuántos años tarda un capital en duplicarse al 4 % anual?

````python
capital = 1000
anios = 0
while capital < 2000:
    capital *= 1.04
    anios += 1
print(f"Se duplica en {anios} años: {capital:.2f} €")
````

# 4. Bucle for

Recorre los elementos de una **secuencia** o, en general, de cualquier objeto **iterable** (lista, tupla, cadena, conjunto, diccionario, `range`…), uno a uno. En cada vuelta, la variable del bucle toma el valor del elemento siguiente.

```python
for elemento in iterable:
    # bloque que se ejecuta para cada elemento
```

`for` se usa cuando **se sabe qué hay que recorrer**: es el bucle más habitual en Python.

## 4.1 `range()`

`range` genera una secuencia de enteros. **El final no se incluye.**

| Llamada | Genera |
|---|---|
| `range(fin)` | 0, 1, …, fin − 1 |
| `range(inicio, fin)` | inicio, …, fin − 1 |
| `range(inicio, fin, paso)` | de `paso` en `paso` (el paso puede ser negativo) |

````python
print(list(range(5)))            # [0, 1, 2, 3, 4]
print(list(range(2, 8)))         # [2, 3, 4, 5, 6, 7]
print(list(range(1, 11, 2)))     # impares del 1 al 9
print(list(range(10, 0, -3)))    # [10, 7, 4, 1]
````

````python
for i in range(1, 11):           # los números del 1 al 10
    print(i, end=" ")
````

> `list(range(…))` solo se usa aquí para **ver** los números. En un bucle se escribe `range` directamente: genera los números según se necesitan, sin crear la lista entera.

## 4.2 `break`: salir del bucle

Termina el bucle **inmediatamente**, aunque queden elementos por recorrer. Se usa sobre todo para **buscar**: en cuanto se encuentra lo que se busca, no hace falta seguir.

````python
for i in range(1, 11):
    if i == 6:
        break             # sale del bucle cuando i vale 6
    print(i, end=" ")
````

````python
importes = [120, 45, 300, -15, 80]
for importe in importes:
    if importe < 0:
        print("Primer importe negativo encontrado:", importe)
        break
````

## 4.3 `continue`: saltar a la siguiente vuelta

Abandona la vuelta actual y pasa **directamente a la siguiente**.

````python
for i in range(1, 11):
    if i % 2 == 0:
        continue          # salta los pares
    print(i, end=" ")     # solo llegan los impares
````

> No conviene abusar de `break` y `continue`: un bucle con varias salidas es difícil de seguir. Muchas veces se puede escribir mejor con una condición. El ejemplo anterior queda más claro así:

````python
for i in range(1, 11):
    if i % 2 != 0:
        print(i, end=" ")
````

## 4.4 `else` en los bucles

Un `for` (o un `while`) puede llevar un bloque `else`, que se ejecuta **solo si el bucle termina sin `break`**. Es útil en las búsquedas: «si no se ha encontrado…».

````python
for n in [7, 9]:
    for divisor in range(2, n):
        if n % divisor == 0:
            print(n, "no es primo")
            break
    else:                       # sin break: no se encontró ningún divisor
        print(n, "es primo")
````

## 4.5 Bucles anidados

Un bucle dentro de otro: el bucle interior se recorre **entero en cada vuelta** del exterior.

````python
for fila in range(1, 4):
    for columna in range(1, 4):
        print(f"{fila * columna:3}", end="")
    print()                    # salto de línea al terminar cada fila
````

# 5. Recorrer colecciones

## 5.1 Listas y tuplas

Se recorren en orden, elemento a elemento:

````python
ciudades = ["Madrid", "Barcelona", "Valencia"]
for ciudad in ciudades:
    print(ciudad)
````

````python
tupla = (1, 2, 3)
for numero in tupla:
    print(numero)
````

## 5.2 Cadenas

Una cadena se recorre **carácter a carácter**:

````python
for letra in "León":
    print(letra, end=" - ")
````

## 5.3 Conjuntos

Se recorren igual, pero **el orden no está garantizado**: un conjunto no tiene orden. Si se necesita un orden, se recorre `sorted(conjunto)`.

````python
provincias = {"Soria", "León", "Ávila"}
for p in sorted(provincias):
    print(p)          # Ávila sale la última: las letras con tilde se ordenan después de las demás
````

## 5.4 Índice y valor: `enumerate()`

Si además del elemento se necesita su **posición**, se usa `enumerate()`, que devuelve en cada vuelta una pareja (índice, valor). Con `start` se elige el número inicial.

````python
ciudades = ["Madrid", "Barcelona", "Valencia"]
for indice, ciudad in enumerate(ciudades):
    print(f"Ciudad {indice}: {ciudad}")

for posicion, ciudad in enumerate(ciudades, start=1):
    print(f"{posicion}.º {ciudad}")
````

> Evita recorrer con `for i in range(len(lista))` para después escribir `lista[i]`: funciona, pero `enumerate` es más claro y más «Python».

## 5.5 Recorrer varias colecciones a la vez: `zip()`

`zip()` empareja los elementos de dos o más colecciones por su posición. Si tienen distinta longitud, se detiene en la más corta.

````python
tiendas = ["León", "Burgos", "Soria"]
ventas = [540, 610, 220]
for tienda, venta in zip(tiendas, ventas):
    print(f"{tienda:<8}{venta:>6} €")
````

## 5.6 Diccionarios

Un diccionario se puede recorrer por sus **claves**, por sus **valores** o por **ambos**. Los diccionarios conservan el **orden en que se insertaron** los elementos.

**Las claves**, recorriendo directamente el diccionario o con `keys()`:

````python
capitales = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}

for pais in capitales:
    print(pais)

for pais in capitales.keys():      # equivalente
    print(pais)
````

**Los valores**, con `values()`:

````python
for ciudad in capitales.values():
    print(ciudad)
````

**Claves y valores a la vez**, con `items()`. Es la forma más habitual:

````python
for pais, ciudad in capitales.items():
    print(f"La capital de {pais} es {ciudad}")
````

> **Ojo.** No se puede añadir ni borrar claves de un diccionario **mientras se recorre** (da `RuntimeError`). Si hace falta, se recorre una copia: `for clave in list(diccionario):`.

## 5.7 Recorrer en otro orden: `sorted()` y `reversed()`

`sorted()` devuelve los elementos ordenados (con `reverse=True`, de mayor a menor) y `reversed()` los recorre al revés. Ninguna de las dos modifica la colección original.

````python
ventas = {"León": 540, "Burgos": 310, "Soria": 620}

for tienda in sorted(ventas):                       # por orden alfabético de la clave
    print(tienda, ventas[tienda])

print()
for tienda, venta in sorted(ventas.items(), key=lambda par: par[1], reverse=True):
    print(tienda, venta)                            # de mayor a menor venta
````

> `key=lambda par: par[1]` indica **por qué** se ordena: por el segundo elemento de cada pareja (la venta). Las funciones `lambda` se ven en los apuntes de funciones.

````python
for i in reversed(range(1, 6)):
    print(i, end=" ")
````

# 6. Patrones habituales con bucles

Casi todos los bucles de un análisis siguen uno de estos esquemas.

## 6.1 Acumulador: sumar

Una variable que empieza en 0 y a la que se va sumando en cada vuelta.

````python
importes = [120.5, 45.0, 300.25, 80.0]
total = 0
for importe in importes:
    total += importe
print("Total:", total)
print("Media:", total / len(importes))
````

## 6.2 Contador: contar los que cumplen algo

````python
edades = [23, 45, 67, 31, 52, 19, 74]
mayores_50 = 0
for edad in edades:
    if edad > 50:
        mayores_50 += 1
print("Mayores de 50:", mayores_50)
````

## 6.3 Máximo y mínimo: buscar el mejor

````python
tiendas = ["León", "Burgos", "Soria", "Ávila"]
ventas = [540, 610, 220, 305]

mejor_tienda = tiendas[0]
mejor_venta = ventas[0]
for tienda, venta in zip(tiendas, ventas):
    if venta > mejor_venta:
        mejor_tienda, mejor_venta = tienda, venta
print("Tienda con más ventas:", mejor_tienda, mejor_venta)
````

## 6.4 Filtrar: construir una lista nueva

````python
importes = [120, -15, 300, 0, 80, -40]
positivos = []
for importe in importes:
    if importe > 0:
        positivos.append(importe)
print(positivos)
````

## 6.5 Contar por categorías con un diccionario

Es el patrón de una **tabla de frecuencias**: el diccionario guarda, para cada valor, cuántas veces ha aparecido. `dict.get(clave, 0)` devuelve 0 si la clave todavía no existe.

````python
provincias = ["León", "Soria", "León", "Ávila", "León", "Soria"]
frecuencias = {}
for p in provincias:
    frecuencias[p] = frecuencias.get(p, 0) + 1
print(frecuencias)
````

## 6.6 Pedir un dato hasta que sea válido

Un uso típico de `while` con `input()`: repetir la pregunta mientras la respuesta no sea correcta. **Esta celda espera a que escribas.**

````python
nota = -1
while not (0 <= nota <= 10):
    texto = input("Introduce una nota entre 0 y 10: ")
    if texto.isdigit():
        nota = int(texto)
    else:
        print("Eso no es un número entero")
print("Nota registrada:", nota)
````

## 6.7 Funciones que ya hacen el bucle

Para los casos más comunes, Python tiene funciones que hacen el bucle por ti, más cortas y más rápidas:

| Función | Qué hace |
|---|---|
| `sum(colección)` | Suma |
| `len(colección)` | Número de elementos |
| `min(colección)`, `max(colección)` | Mínimo y máximo |
| `sorted(colección)` | Lista ordenada |
| `any(condiciones)` | ¿Se cumple alguna? |
| `all(condiciones)` | ¿Se cumplen todas? |

````python
importes = [120.5, 45.0, 300.25, 80.0]
print(sum(importes), len(importes), min(importes), max(importes))
print(sum(importes) / len(importes))
print(any(x > 200 for x in importes))        # ¿alguno supera 200?
print(all(x > 0 for x in importes))          # ¿son todos positivos?
````

# 7. Comprensiones

Una **comprensión** construye una lista (o un diccionario, o un conjunto) a partir de otra colección **en una sola línea** mediante una sintaxis especial. Es la forma habitual en Python de escribir los patrones de transformar y filtrar.

## 7.1 Comprensión de listas

```python
# Sintaxis para crear una lista
nueva = [expresión for elemento in colección]
# Sintaxis para crear una lista filtrada, mediante una condición
nueva = [expresión for elemento in colección if condición]
```

````python
numeros = [1, 2, 3, 4, 5]
# Transformar
cuadrados = [n ** 2 for n in numeros] # Lista con los cuadrados de los números del 1 al 5
print(cuadrados)

# Filtrar
importes = [120, -15, 300, 0, 80, -40]
positivos = [x for x in importes if x > 0]  # Lista con los importes positivos
print(positivos)

# Filtrar y transformar
con_iva = [round(x * 1.21, 2) for x in importes if x > 0]   # Lista con los importes positivos con IVA
print(con_iva)
````

Es equivalente al patrón del apartado 6.4, en una línea:

```python
positivos = []                         
positivos = [x for x in importes if x > 0]

for x in importes:
    if x > 0:
        positivos.append(x)
```

## 7.2 Comprensión de diccionarios

````python
provincias = ["León", "Soria", "Ávila"]
longitudes = {p: len(p) for p in provincias} # Diccionario con la longitud de cada provincia
print(longitudes)

precios = {"pan": 1.20, "leche": 0.95, "aceite": 8.50}
con_iva = {producto: round(precio * 1.10, 2) for producto, precio in precios.items()} # Diccionario con los precios con IVA
print(con_iva)
````

> Las comprensiones se leen bien mientras son cortas. Si una comprensión necesita varias condiciones o varias líneas, un bucle normal es más claro.

# 8. Manejo de excepciones

Una **excepción** es un error que se produce **durante la ejecución**: convertir un texto que no es un número, dividir entre cero, abrir un fichero que no existe… Si no se gestiona, **detiene el programa**. La estructura `try` permite capturarla y decidir qué hacer.

## 8.1 `try-except`

```python
try:
    # código que puede producir una excepción
except TipoDeExcepcion:
    # qué hacer si se produce
```

Si dentro del `try` se produce la excepción indicada, Python salta al bloque `except` y continúa después. Si no se produce, el `except` no se ejecuta.

````python
for texto in ["5", "0", "hola"]:
    try:
        numero = int(texto)
        resultado = 10 / numero
        print(f"10 / {numero} = {resultado}")
    except ValueError:                   # errores de conversión
        print(f"'{texto}' no es un número válido")
    except ZeroDivisionError:            # división entre cero
        print("No se puede dividir entre cero")
````

## 8.2 El mensaje del error: `as`

Con `as` se guarda la excepción en una variable para mostrar su mensaje. Varios tipos se pueden capturar juntos con una **tupla**.

````python
for texto in ["0", "hola"]:
    try:
        print(10 / int(texto))
    except (ValueError, ZeroDivisionError) as e:
        print(f"Error de tipo {type(e).__name__}: {e}")
````

## 8.3 `else` y `finally`

* **`else`**: se ejecuta solo si **no** hubo excepción. Sirve para dejar en el `try` únicamente la instrucción que puede fallar.
* **`finally`**: se ejecuta **siempre**, haya error o no. Se usa para cerrar o liberar recursos (ficheros, conexiones a bases de datos).

````python
for texto in ["4", "cuatro"]:
    try:
        numero = int(texto)
    except ValueError:
        print(f"'{texto}': conversión fallida")
    else:
        print(f"'{texto}': conversión correcta, el doble es {numero * 2}")
    finally:
        print("  (fin del intento)")
````

## 8.4 Excepciones frecuentes

| Excepción | Cuándo se produce |
|---|---|
| `ValueError` | Valor del tipo correcto pero inválido: `int("hola")` |
| `TypeError` | Operación con tipos incompatibles: `"a" + 1` |
| `ZeroDivisionError` | División entre cero |
| `IndexError` | Índice fuera de rango en una lista: `lista[10]` |
| `KeyError` | Clave inexistente en un diccionario |
| `NameError` | Variable no definida |
| `FileNotFoundError` | El fichero no existe |
| `ModuleNotFoundError` | La librería no está instalada |

## 8.5 La excepción genérica `Exception`

`except Exception` captura **cualquier** excepción. Es útil como último recurso, pero conviene capturar las excepciones **concretas**: así se trata cada error como corresponde y no se esconden errores que no se esperaban.

Si se usan varias, **`Exception` va la última**: los `except` se comprueban en orden, y si fuera la primera capturaría todo y los demás nunca se ejecutarían.

````python
datos = {"León": 540}
for clave in ["León", "Soria"]:
    try:
        print(clave, datos[clave] / 2)
    except KeyError:
        print(f"No hay datos de {clave}")
    except Exception as e:               # cualquier otro error, al final
        print("Error inesperado:", e)
````

> **No uses `except:` sin indicar tipo**, ni `except Exception: pass`. Esconden todos los errores, también los que delatan un fallo en tu código, y el programa sigue con datos incorrectos sin avisar.

## 8.6 Lanzar una excepción: `raise`

Un programa puede **provocar** una excepción cuando detecta una situación que no debe continuar:

````python
edad = -5
try:
    if edad < 0:
        raise ValueError(f"La edad no puede ser negativa: {edad}")
except ValueError as e:
    print("ValueError:", e)
````

> Para **comprobar** que los datos son correctos en un análisis se usa `assert condición, "mensaje"`, que lanza un `AssertionError` si la condición es falsa. Se trabaja en los apuntes de NumPy.

## 8.7 Un ejemplo con `input()`

**Esta celda espera a que escribas un número.** Prueba con un número, con 0 y con un texto.

````python
try:
    numero = int(input("Introduce un número: "))
    resultado = 10 / numero
except ValueError:
    print("Error: debes introducir un número entero.")
except ZeroDivisionError:
    print("Error: no se puede dividir entre cero.")
else:
    print(f"El resultado es: {resultado}")
````

# 9. Bucles y análisis de datos

Los bucles son imprescindibles para aprender a programar, pero **en análisis de datos se evitan** siempre que se puede: con NumPy y pandas, las operaciones se aplican a **columnas enteras de una vez** (*vectorización*), lo que es mucho más rápido y más corto.

````python
importes = [100.0, 250.0, 80.0, 40.0]

con_iva = []                                   # con un bucle
for x in importes:
    con_iva.append(round(x * 1.21, 2))
print(con_iva)

import numpy as np                             # con NumPy, sin bucle
print(np.round(np.array(importes) * 1.21, 2))
````

Lo que sí sigue siendo habitual en un análisis: recorrer **ficheros**, **columnas** o **grupos** para repetir el mismo proceso con cada uno, y los `try`/`except` al leer datos que pueden venir mal.

# 10. Errores frecuentes

| Lo que ves | Qué pasa | Solución |
|---|---|---|
| `SyntaxError: expected ':'` | Falta el `:` al final del `if`, `for`, `while`… | Añade los dos puntos |
| `IndentationError` | La sangría no es coherente, o falta el bloque | 4 espacios por nivel; `pass` si el bloque va vacío |
| Una instrucción se ejecuta siempre (o nunca) | Está a la sangría equivocada: dentro o fuera del bloque | Revisa la sangría |
| La celda se queda en `[*]` | Bucle infinito: la condición del `while` nunca cambia | *Kernel → Interrupt Kernel*; revisa la variable de la condición |
| `if x = 5:` da `SyntaxError` | `=` asigna; para comparar es `==` | `if x == 5:` |
| Un `elif` nunca se ejecuta | Una condición anterior más general ya lo cubre | Ordena de la más restrictiva a la menos |
| `range(1, 10)` no incluye el 10 | El final de `range` no se incluye | `range(1, 11)` |
| `RuntimeError: dictionary changed size during iteration` | Se modifica el diccionario mientras se recorre | Recorre `list(diccionario)` |
| Un error desaparece sin explicación | Un `except` demasiado general lo esconde | Captura excepciones concretas |
