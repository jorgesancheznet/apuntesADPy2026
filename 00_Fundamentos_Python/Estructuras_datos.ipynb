# 1. Visión general

| | Lista | Tupla | Conjunto | Diccionario |
|---|---|---|---|---|
| Sintaxis | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` | `{"a": 1, "b": 2}` |
| Vacío | `[]` o `list()` | `()` o `tuple()` | **solo** `set()` | `{}` o `dict()` |
| ¿Ordenada? | Sí (por posición) | Sí (por posición) | **No** | Sí (orden de inserción) |
| ¿Se puede modificar? | Sí | **No** | Sí | Sí |
| ¿Admite duplicados? | Sí | Sí | **No** | Claves no; valores sí |
| Acceso | `lista[0]` | `tupla[0]` | No hay acceso por posición | `dic["a"]` |
| Uso típico | Secuencia de datos que cambia | Registro fijo: coordenadas, fecha | Valores únicos, pertenencia | Buscar por clave: ficha, tabla |

Las cuatro comparten algunas operaciones: `len()` da el número de elementos, `in` comprueba si un elemento está, y `for` las recorre.

````python
lista = [10, 20, 30]
tupla = (10, 20, 30)
conjunto = {10, 20, 30}
diccionario = {"a": 10, "b": 20, "c": 30}

for estructura in (lista, tupla, conjunto, diccionario):
    print(type(estructura).__name__, len(estructura), 20 in estructura)
````

> En el diccionario, `20 in diccionario` da `False` porque `in` busca entre las **claves** (`"a"`, `"b"`, `"c"`), no entre los valores. Lo verás en el apartado 5.

# 2. Listas

## 2.1 ¿Qué es una lista?

Una lista es una **secuencia ordenada y modificable** de elementos. Se escribe entre corchetes `[]`, con los elementos separados por comas. Puede mezclar tipos e incluso contener otras listas, aunque en análisis de datos lo normal es que todos los elementos sean del mismo tipo.

## 2.2 Creación de listas

````python
vacia = []                               # lista vacía
numeros = [1, 2, 3, 4, 5]                # números
letras = ["a", "b", "c", "d"]            # cadenas
mezcla = [1, "b", 3.5, True]             # tipos distintos
anidada = ["Hola", 18, 25, [2, 3, 4]]    # con una sublista
print(vacia)
print(numeros)
print(letras)
print(mezcla)
print(anidada)
````

También se crean con la función `list()`, a partir de cualquier cosa que se pueda recorrer:

````python
print(list())                  # []
print(list((1, 2, 3)))         # [1, 2, 3]            desde una tupla
print(list("Hola"))            # ['H', 'o', 'l', 'a'] desde una cadena
print(list(range(5)))          # [0, 1, 2, 3, 4]      desde un rango
````

## 2.3 Acceso a los elementos

Cada elemento tiene una **posición** (índice) que empieza en **0**. Los índices negativos cuentan desde el final: `-1` es el último.

```
 lista =  [ 10,  20,  30,  40,  50 ]
 índice:     0    1    2    3    4
 negativo:  -5   -4   -3   -2   -1
```

````python
lista = [10, 20, 30, 40, 50]
print(lista[0])     # 10  primer elemento
print(lista[2])     # 30  tercer elemento
print(lista[-1])    # 50  último elemento
print(lista[-2])    # 40  penúltimo
````

Pedir una posición que no existe produce un `IndexError`. Una lista de 5 elementos tiene índices del 0 al 4:

````python
lista = [10, 20, 30, 40, 50]
try:
    print(lista[5])
except IndexError as e:
    print("IndexError:", e)
````

## 2.4 Rebanadas (*slicing*)

Una rebanada obtiene una **sublista** con la sintaxis `lista[inicio:fin:paso]`:

- `inicio`: posición donde empieza (**incluida**). Si se omite, desde el principio.
- `fin`: posición donde termina (**no incluida**). Si se omite, hasta el final.
- `paso`: de cuántos en cuántos (opcional, por defecto 1). Si es negativo, recorre al revés.

Una rebanada **nunca** da `IndexError`: si los límites se salen, se queda con lo que haya.

````python
lista = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(lista[2:5])      # [30, 40, 50]
print(lista[:4])       # [10, 20, 30, 40]    los 4 primeros
print(lista[5:])       # [60, 70, 80, 90]    desde la posición 5
print(lista[-3:])      # [70, 80, 90]        los 3 últimos
print(lista[::2])      # [10, 30, 50, 70, 90]
print(lista[1:8:3])    # [20, 50, 80]
print(lista[::-1])     # [90, 80, 70, 60, 50, 40, 30, 20, 10]  al revés
print(lista[7:100])    # [80, 90]            sin error
````

## 2.5 Modificar elementos

Las listas son **mutables**: se puede cambiar un elemento asignándole un valor nuevo en su posición, o cambiar un tramo entero con una rebanada.

````python
lista = [10, 20, 30, 40, 50]
lista[1] = 25                  # cambia el segundo elemento
print(lista)                   # [10, 25, 30, 40, 50]

lista[2:4] = [0, 0]            # cambia un tramo
print(lista)                   # [10, 25, 0, 0, 50]
````

## 2.6 Tamaño y funciones de resumen

`len()` da el número de elementos. Si la lista es de números, `sum()`, `min()` y `max()` dan el total, el mínimo y el máximo.

````python
ventas = [1200, 850, 1430, 990, 1610]
print("Elementos:", len(ventas))                      # 5
print("Total:", sum(ventas))                          # 6080
print("Mínimo:", min(ventas), "Máximo:", max(ventas)) # 850 1610
print("Media:", sum(ventas) / len(ventas))            # 1216.0
````

## 2.7 Concatenación y repetición

- `+` une dos listas en una nueva.
- `*` repite una lista un número de veces.

Ninguno de los dos modifica las listas originales.

````python
a = [10, 20, 30]
b = [40, 50]
print(a + b)        # [10, 20, 30, 40, 50]
print(a * 2)        # [10, 20, 30, 10, 20, 30]
print(a)            # [10, 20, 30]  a no cambia
````

## 2.8 Añadir elementos

| Método | Qué hace |
|---|---|
| `append(x)` | Añade **un** elemento al final |
| `insert(i, x)` | Inserta `x` en la posición `i` y desplaza el resto |
| `extend(otra)` | Añade **todos** los elementos de otra colección al final |

Los tres **modifican la lista** y devuelven `None`.

````python
lista = [10, 20, 30]
lista.append(40)
print(lista)              # [10, 20, 30, 40]

lista.insert(0, 5)        # en la posición 0
print(lista)              # [5, 10, 20, 30, 40]

lista.extend([50, 60])
print(lista)              # [5, 10, 20, 30, 40, 50, 60]
````

**`append` frente a `extend`.** `append` mete lo que le des como **un solo** elemento, aunque sea una lista; `extend` mete sus elementos **uno a uno**:

````python
a = [1, 2]
a.append([3, 4])
print(a, len(a))      # [1, 2, [3, 4]] 3

b = [1, 2]
b.extend([3, 4])
print(b, len(b))      # [1, 2, 3, 4] 4
````

## 2.9 Eliminar elementos

| Forma | Qué hace | Devuelve |
|---|---|---|
| `remove(x)` | Quita la **primera** aparición del **valor** `x`. Si no está, `ValueError` | `None` |
| `pop(i)` | Quita el elemento de la **posición** `i` | El elemento quitado |
| `pop()` | Quita el **último** | El elemento quitado |
| `del lista[i]` | Quita la posición `i` (o una rebanada). Es una **instrucción**, no un método | — |
| `clear()` | Vacía la lista | `None` |

````python
lista = [10, 20, 30, 20, 40, 20, 50]
lista.remove(20)          # quita el primer 20
print(lista)              # [10, 30, 20, 40, 20, 50]

try:
    lista.remove(99)
except ValueError as e:
    print("ValueError:", e)
````

````python
lista = [10, 20, 30, 40, 50]
elemento = lista.pop(2)   # posición 2
print("Eliminado:", elemento, "->", lista)   # 30 -> [10, 20, 40, 50]

ultimo = lista.pop()      # sin posición: el último
print("Eliminado:", ultimo, "->", lista)     # 50 -> [10, 20, 40]
````

````python
lista = [10, 20, 30, 40, 50]
del lista[0]
print(lista)              # [20, 30, 40, 50]
del lista[1:3]            # también con rebanadas
print(lista)              # [20, 50]

lista.clear()
print(lista)              # []
````

## 2.10 Buscar y contar

- `x in lista` devuelve `True` o `False`. Es la forma segura de preguntar si algo está.
- `lista.index(x)` devuelve la posición de la **primera** aparición. Si no está, `ValueError`.
- `lista.count(x)` cuenta cuántas veces aparece.

````python
lista = [10, 20, 30, 20, 40]
print(20 in lista)        # True
print(60 in lista)        # False
print(60 not in lista)    # True
print(lista.index(20))    # 1
print(lista.count(20))    # 2

try:
    print(lista.index(70))
except ValueError as e:
    print("ValueError:", e)
````

Para buscar sin riesgo de error, primero se pregunta con `in`:

````python
lista = [10, 20, 30, 20, 40]
buscado = 70
if buscado in lista:
    print("Está en la posición", lista.index(buscado))
else:
    print(buscado, "no está en la lista")
````

## 2.11 Ordenar e invertir

Hay **dos** formas de ordenar, y confundirlas es un error muy común:

| | `lista.sort()` | `sorted(lista)` |
|---|---|---|
| ¿Qué es? | Método de las listas | Función que sirve para cualquier colección |
| ¿Modifica la lista? | **Sí** | No |
| ¿Qué devuelve? | **`None`** | Una **lista nueva** ordenada |

Las dos admiten `reverse=True` (de mayor a menor) y `key=` (ordenar según una función).

````python
lista = [20, 40, 10, 30, 60, 50]
ordenada = sorted(lista)
print(ordenada)           # [10, 20, 30, 40, 50, 60]
print(lista)              # [20, 40, 10, 30, 60, 50]  sin cambios

lista.sort(reverse=True)
print(lista)              # [60, 50, 40, 30, 20, 10]  modificada
````

**La trampa:** como `sort()` devuelve `None`, guardar su resultado deja la variable vacía.

````python
lista = [3, 1, 2]
resultado = lista.sort()
print(resultado)          # None
print(lista)              # [1, 2, 3]
````

Con `key=` se ordena según un criterio. Por ejemplo, palabras por su longitud o sin distinguir mayúsculas:

````python
ciudades = ["León", "Ávila", "burgos", "Soria", "Zamora"]
print(sorted(ciudades))                   # las minúsculas y las tildes van después
print(sorted(ciudades, key=str.lower))    # sin distinguir mayúsculas
print(sorted(ciudades, key=len))          # por número de letras
````

> Python ordena las cadenas por el código de cada carácter: las mayúsculas van antes que las minúsculas y las letras con tilde, después de la `z`. Por eso `"burgos"` y `"Ávila"` quedan al final en el primer caso. `key=str.lower` arregla las mayúsculas, pero no las tildes.

`reverse()` invierte el orden de la lista (la modifica, como `sort()`). La rebanada `[::-1]` hace lo mismo sin modificarla:

````python
lista = [1, 2, 3, 4]
print(lista[::-1])        # [4, 3, 2, 1]  copia invertida
lista.reverse()
print(lista)              # [4, 3, 2, 1]  la lista cambia
````

## 2.12 Listas anidadas

Una lista puede contener listas. Es la forma más simple de representar una **tabla**: cada sublista es una fila. Se accede con dos índices: `tabla[fila][columna]`.

````python
# ventas (miles de €) de 3 tiendas en 4 trimestres
tabla = [
    [120, 135, 150, 170],
    [ 90,  95, 110, 100],
    [200, 210, 190, 230],
]
print(tabla[0])           # [120, 135, 150, 170]  fila 0: primera tienda
print(tabla[2][3])        # 230  tienda 3, trimestre 4
print(len(tabla), "filas y", len(tabla[0]), "columnas")

for fila in tabla:
    print(sum(fila))      # total anual de cada tienda
````

> **Cuidado al crear tablas vacías.** `[[0] * 3] * 3` **no** crea tres filas independientes, sino tres referencias a **la misma** fila. Al cambiar una, cambian todas (se explica en el apartado 7). La forma correcta es con una comprensión:

````python
mal = [[0] * 3] * 3
mal[0][0] = 1
print(mal)        # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]  ¡las tres filas!

bien = [[0] * 3 for _ in range(3)]
bien[0][0] = 1
print(bien)       # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
````

# 3. Tuplas

## 3.1 ¿Qué es una tupla?

Una tupla es una secuencia ordenada como la lista, pero **inmutable**: una vez creada, no se pueden cambiar, añadir ni quitar elementos. Se escribe entre paréntesis `()`.

Se usan para datos que forman un **registro fijo**: unas coordenadas `(latitud, longitud)`, una fecha `(año, mes, día)` o un par `(clave, valor)`. Al no poder modificarse, sirven como claves de diccionario y como elementos de conjuntos, cosa que las listas no pueden.

## 3.2 Creación de tuplas

````python
tupla = (10, 20, 30, 40, 50)
print(tupla)

coordenadas = 42.60, -5.57        # los paréntesis son opcionales: lo que crea la tupla es la coma
print(coordenadas, type(coordenadas))

vacia = ()
print(vacia, len(vacia))
print(tuple([1, 2, 3]))           # desde una lista
````

**La tupla de un elemento** necesita una coma al final. Sin ella, los paréntesis son solo paréntesis de una expresión:

````python
no_es_tupla = (5)
si_es_tupla = (5,)
print(type(no_es_tupla))    # <class 'int'>
print(type(si_es_tupla))    # <class 'tuple'>
````

## 3.3 Acceso y rebanadas

Funcionan exactamente igual que en las listas, y la rebanada de una tupla es otra tupla.

````python
tupla = (10, 20, 30, 40, 50)
print(tupla[0])       # 10
print(tupla[-1])      # 50
print(tupla[-2])      # 40
print(tupla[1:4])     # (20, 30, 40)
print(tupla[::2])     # (10, 30, 50)
````

## 3.4 Inmutabilidad

Intentar modificar una tupla produce un `TypeError`. Tampoco tiene `append`, `remove` ni `sort`.

````python
tupla = (10, 20, 30)
try:
    tupla[0] = 99
except TypeError as e:
    print("TypeError:", e)

try:
    tupla.append(40)
except AttributeError as e:
    print("AttributeError:", e)
````

Para "cambiar" una tupla se crea otra nueva, por ejemplo concatenando o pasando por una lista:

````python
tupla = (10, 20, 30)
tupla = tupla + (40,)          # nueva tupla asignada a la misma variable
print(tupla)                   # (10, 20, 30, 40)

temporal = list(tupla)
temporal[0] = 99
tupla = tuple(temporal)
print(tupla)                   # (99, 20, 30, 40)
````

## 3.5 Operaciones y métodos

- `len()`, `in`, `min()`, `max()`, `sum()` y `sorted()` funcionan igual que con listas (`sorted` devuelve una **lista**).
- `+` concatena dos tuplas y `*` repite una tupla.
- Solo tiene dos métodos: `count(x)`, que cuenta las apariciones de `x`, e `index(x)`, que da la posición de la primera aparición de `x` en la tupla.

````python
tupla = (30, 10, 20, 10, 40)
print(len(tupla), 20 in tupla, sum(tupla))   # 5 True 110
print(tupla + (50, 60))                      # (30, 10, 20, 10, 40, 50, 60)
print((1, 2) * 3)                            # (1, 2, 1, 2, 1, 2)
print(tupla.count(10))                       # 2
print(tupla.index(20))                       # 2
print(sorted(tupla))                         # [10, 10, 20, 30, 40]  es una lista
````

## 3.6 Desempaquetado

El desempaquetado asigna cada elemento de una tupla (o de una lista) a una variable distinta en una sola línea. Tiene que haber **tantas variables como elementos**, salvo que una lleve `*`, que recoge "el resto" en una lista.

````python
punto = (42.60, -5.57)
latitud, longitud = punto
print(latitud)                  # 42.6
print(longitud)                 # -5.57

primero, *resto = [10, 20, 30, 40]
print(primero, resto)           # 10 [20, 30, 40]

try:
    a, b = (1, 2, 3)
except ValueError as e:
    print("ValueError:", e)
````

Dos usos muy habituales: **intercambiar** dos variables sin auxiliar, y recoger los **varios resultados** que devuelve una función (que llegan como tupla):

````python
a, b = 1, 2
a, b = b, a
print(a, b)                     # 2 1

cociente, resto = divmod(17, 5) # divmod devuelve una tupla (cociente, resto)
print(cociente, resto)          # 3 2
````

# 4. Conjuntos

## 4.1 ¿Qué es un conjunto?

Un conjunto (`set`) es una colección **sin orden** y **sin elementos repetidos**. Es útil para quedarse con los valores únicos, para comprobar muy rápido si un valor está y para las operaciones de la teoría de conjuntos: unión, intersección y diferencia.

## 4.2 Creación de conjuntos

Se escriben entre llaves `{}` o con la función `set()`. Los duplicados se eliminan solos.

````python
conjunto = {20, 40, 60, 40, 80}
print(conjunto, len(conjunto))      # 4 elementos: el 40 repetido desaparece

print(set([3, 1, 3, 2, 1]))         # desde una lista
print(set("banana"))                # letras únicas, en cualquier orden
````

> **El orden en que se muestran los elementos no está garantizado.** Puede no coincidir con el de creación, e incluso cambiar de una ejecución a otra (pasa con los conjuntos de cadenas). No hay que contar con él: si se necesita un orden, se usa `sorted(conjunto)`, que devuelve una lista ordenada.

**El conjunto vacío se crea con `set()`**, no con `{}`: las llaves vacías crean un **diccionario** vacío.

````python
print(type({}))       # <class 'dict'>
print(type(set()))    # <class 'set'>
````

## 4.3 Lo que se puede y no se puede hacer

- **No** se accede por posición (`conjunto[0]` da `TypeError`) ni se hacen rebanadas, porque no hay orden.
- **Sí** se puede usar `in`: es la operación en la que los conjuntos son **más rápidos** que las listas.
- Sus elementos tienen que ser **inmutables** (números, cadenas, tuplas). Una lista no puede ser elemento de un conjunto.

````python
provincias = {"León", "Burgos", "Soria"}
print("León" in provincias)       # True
print("Lugo" in provincias)       # False

try:
    print(provincias[0])
except TypeError as e:
    print("TypeError:", e)

try:
    malo = {[1, 2], [3, 4]}
except TypeError as e:
    print("TypeError:", e)
````

## 4.4 Añadir y eliminar elementos

| Método | Qué hace |
|---|---|
| `add(x)` | Añade `x` (si ya estaba, no pasa nada) |
| `remove(x)` | Quita `x`. Si no está, `KeyError` |
| `discard(x)` | Quita `x` si está. Si no está, **no** da error |
| `clear()` | Vacía el conjunto |

````python
numeros = {1, 2, 3}
numeros.add(4)
numeros.add(2)                    # ya estaba: no cambia
print(sorted(numeros))            # [1, 2, 3, 4]

numeros.remove(1)
numeros.discard(99)               # no está, pero no da error
print(sorted(numeros))            # [2, 3, 4]

try:
    numeros.remove(99)
except KeyError as e:
    print("KeyError:", e)
````

## 4.5 Operaciones con conjuntos

Cada operación tiene un operador y un método equivalente. Ninguno modifica los conjuntos originales.

| Operación | Operador | Método | Resultado |
|---|---|---|---|
| Unión | `a \| b` | `a.union(b)` | Lo que está en `a` **o** en `b` |
| Intersección | `a & b` | `a.intersection(b)` | Lo que está en `a` **y** en `b` |
| Diferencia | `a - b` | `a.difference(b)` | Lo que está en `a` pero **no** en `b` |
| Diferencia simétrica | `a ^ b` | `a.symmetric_difference(b)` | Lo que está en uno solo de los dos |

````python
a = {20, 40, 60}
b = {30, 40, 50, 60}
print(sorted(a | b))      # [20, 30, 40, 50, 60]
print(sorted(a & b))      # [40, 60]
print(sorted(a - b))      # [20]
print(sorted(b - a))      # [30, 50]   la diferencia no es simétrica
print(sorted(a ^ b))      # [20, 30, 50]        # <class 'set'>  la diferencia simétrica es un conjunto

print(a.union(b) == a | b)            # True: mismo resultado
````

Un ejemplo con datos: qué clientes compraron en los dos meses, cuáles son nuevos y cuáles se han perdido.

````python
enero = {"C01", "C02", "C03", "C05"}
febrero = {"C02", "C03", "C04", "C06"}
print("Repiten:", sorted(enero & febrero))      # ['C02', 'C03']
print("Nuevos:", sorted(febrero - enero))       # ['C04', 'C06']
print("Perdidos:", sorted(enero - febrero))     # ['C01', 'C05']
print("Total distintos:", len(enero | febrero)) # 6
````

## 4.6 Subconjuntos y conjuntos disjuntos

- `a <= b` (o `a.issubset(b)`): ¿están todos los elementos de `a` en `b`?
- `a >= b` (o `a.issuperset(b)`): ¿contiene `a` a todos los de `b`?
- `a.isdisjoint(b)`: ¿no tienen **ningún** elemento en común?

````python
obligatorias = {"fecha", "tienda", "importe"}
columnas = {"fecha", "tienda", "importe", "cliente"}
print(obligatorias <= columnas)            # True: el fichero trae todas las columnas
print(columnas >= obligatorias)            # True
print({1, 2}.isdisjoint({3, 4}))           # True
````

## 4.7 Quitar duplicados de una lista

Es el uso más habitual de los conjuntos: `set(lista)` se queda con los valores únicos. Si además se quiere conservar el **orden** de la primera aparición, se usa `dict.fromkeys(lista)`, porque los diccionarios sí conservan el orden (apartado 5).

````python
tiendas = ["T03", "T01", "T03", "T02", "T01", "T03"]
print(len(set(tiendas)), "tiendas distintas")   # 3 tiendas distintas
print(sorted(set(tiendas)))                     # ['T01', 'T02', 'T03']  ordenadas
print(list(dict.fromkeys(tiendas)))             # ['T03', 'T01', 'T02']  en orden de aparición
````

> Existe también `frozenset`, un conjunto **inmutable**. Se usa poco, en los casos en que hace falta un conjunto como clave de diccionario o como elemento de otro conjunto.

# 5. Diccionarios

## 5.1 ¿Qué es un diccionario?

Un diccionario guarda pares **clave → valor**. En lugar de acceder por posición, se accede por **clave**, igual que en un diccionario de papel se busca por la palabra para leer su definición.

- Las **claves** son únicas y tienen que ser de un tipo **inmutable** (técnicamente, *hashable*): números, cadenas, tuplas… Una lista **no** puede ser clave.
- Los **valores** pueden ser de cualquier tipo, repetirse y ser a su vez listas u otros diccionarios.
- Desde Python 3.7, los diccionarios **conservan el orden de inserción**: se recorren y se muestran en el orden en que se añadieron las claves.

## 5.2 Creación de diccionarios

````python
capitales = {
    "España": "Madrid",
    "Francia": "París",
    "Italia": "Roma",
}
print(capitales)

alumnos = {25: "Ana", 30: "Luis", 35: "María"}    # las claves pueden ser números
print(alumnos)

print(dict(España="Madrid", Francia="París"))     # con dict() y nombres
print(dict([("a", 1), ("b", 2)]))                 # desde una lista de pares
vacio = {}
print(vacio, len(vacio))
````

````python
try:
    malo = {["León", "Burgos"]: 2}
except TypeError as e:
    print("TypeError:", e)

bueno = {("León", 2025): 19063385.33}   # una tupla sí vale como clave
print(bueno)
````

## 5.3 Acceso a los valores

- `dic[clave]` devuelve el valor asociado a la clave. Si la clave no existe, `KeyError`.
- `dic.get(clave)` devuelve el valor, o `None` si no existe. `dic.get(clave, defecto)` devuelve `defecto` en lugar de `None`.
- `clave in dic` comprueba si existe la clave (busca en las **claves**, no en los valores).

````python
capitales = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
print(capitales["España"])                  # Madrid

try:
    print(capitales["Portugal"])
except KeyError as e:
    print("KeyError:", e)

print(capitales.get("Portugal"))            # None
print(capitales.get("Portugal", "?"))       # ?
print("Italia" in capitales)                # True
print("Roma" in capitales)                  # False: "Roma" es un valor, no una clave
print("Roma" in capitales.values())         # True
````

## 5.4 Añadir y modificar

La misma sintaxis sirve para las dos cosas: si la clave **no existe**, se añade; si **ya existe**, se sustituye su valor. Por eso un diccionario nunca tiene claves repetidas.

````python
capitales = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
capitales["Alemania"] = "Berlín"            # clave nueva: se añade al final
capitales["Italia"] = "ROMA"                # clave existente: se sustituye
print(capitales)
print(len(capitales))                       # 4
````

`setdefault(clave, valor)` añade la clave con ese valor **solo si no existía**, y en cualquier caso devuelve el valor que queda:

````python
stock = {"tornillos": 100}
stock.setdefault("tuercas", 0)          # no existía: se añade con 0
stock.setdefault("tornillos", 0)        # ya existía: no se toca
print(stock)                            # {'tornillos': 100, 'tuercas': 0}
````

## 5.5 Recorrer un diccionario

| Método | Devuelve |
|---|---|
| `keys()` | Las claves |
| `values()` | Los valores |
| `items()` | Los pares `(clave, valor)` como tuplas |

Recorrer el diccionario directamente (`for k in dic`) equivale a recorrer las claves.

````python
ventas = {"León": 19063, "Burgos": 12880, "Valladolid": 14758}   # miles de € (ejemplo)
print(list(ventas.keys()))
print(list(ventas.values()))
print(list(ventas.items()))

for provincia, importe in ventas.items():
    print(f"{provincia:<12}{importe:>8,}")

print("Total:", sum(ventas.values()))
print("Mayor venta:", max(ventas, key=ventas.get))    # la clave con el valor máximo
````

## 5.6 Eliminar elementos

| Forma | Qué hace | Devuelve |
|---|---|---|
| `pop(clave)` | Quita esa clave. Si no existe, `KeyError` | El valor quitado |
| `pop(clave, defecto)` | Igual, pero si no existe devuelve `defecto` sin error | El valor o `defecto` |
| `popitem()` | Quita el **último par añadido** | La tupla `(clave, valor)` |
| `del dic[clave]` | Quita esa clave. Es una **instrucción**, no un método | — |
| `clear()` | Vacía el diccionario | `None` |

````python
capitales = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
valor = capitales.pop("Francia")
print("Eliminado:", valor)              # París
print(capitales)                        # {'España': 'Madrid', 'Italia': 'Roma'}
print(capitales.pop("Japón", "no estaba"))
````

````python
capitales = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
del capitales["Italia"]
print(capitales)                        # {'España': 'Madrid', 'Francia': 'París'}

capitales["Grecia"] = "Atenas"
print(capitales.popitem())              # ('Grecia', 'Atenas')  el último añadido
print(capitales)                        # {'España': 'Madrid', 'Francia': 'París'}

capitales.clear()
print(capitales)                        # {}
````

## 5.7 Unir diccionarios

- `a | b` crea un diccionario **nuevo** con las claves de los dos.
- `a.update(b)` **modifica** `a` añadiéndole las claves de `b`.

Si una clave está en los dos, gana el valor de `b` (el de la derecha).

````python
capitales1 = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
capitales2 = {"Alemania": "Berlín", "Portugal": "Lisboa", "Italia": "ROMA"}

union = capitales1 | capitales2
print(union)                 # Italia queda con "ROMA"
print(capitales1)            # sin cambios

capitales1.update(capitales2)
print(capitales1)            # ahora sí cambia
````

## 5.8 Contar con un diccionario

Un patrón muy frecuente: recorrer unos datos y llevar la cuenta de cada valor. `get(clave, 0)` evita el `KeyError` la primera vez que aparece una clave.

````python
tickets = ["León", "Burgos", "León", "Soria", "León", "Burgos"]
cuenta = {}
for provincia in tickets:
    cuenta[provincia] = cuenta.get(provincia, 0) + 1
print(cuenta)            # {'León': 3, 'Burgos': 2, 'Soria': 1}
````

> En la UT3 se hará lo mismo en una sola línea con pandas: `serie.value_counts()`.

## 5.9 Diccionarios anidados

Los valores pueden ser listas u otros diccionarios, lo que permite representar fichas completas. Se accede encadenando corchetes.

````python
alumnos = {
    25: {"nombre": "Ana", "edad": 20, "cursos": ["Matemáticas", "Física"]},
    30: {"nombre": "Luis", "edad": 22, "cursos": ["Química", "Biología"]},
    35: {"nombre": "María", "edad": 21, "cursos": ["Historia", "Literatura"]},
}
print(alumnos[30])                      # la ficha completa de Luis
print(alumnos[30]["nombre"])            # Luis
print(alumnos[35]["cursos"][0])         # Historia

alumnos[25]["cursos"].append("Química") # se modifica la lista de dentro
print(alumnos[25]["cursos"])

for numero, ficha in alumnos.items():
    print(numero, ficha["nombre"], len(ficha["cursos"]), "cursos")
````

## 5.10 `dict.fromkeys`

Crea un diccionario con las claves de una colección y el **mismo** valor para todas:

````python
meses = ["ene", "feb", "mar"]
print(dict.fromkeys(meses, 0))          # {'ene': 0, 'feb': 0, 'mar': 0}
````

> No uses un valor mutable (una lista) en `fromkeys`: todas las claves compartirían **la misma** lista. Para eso se usa una comprensión: `{m: [] for m in meses}`.

# 6. Conversiones entre estructuras

Las funciones `list()`, `tuple()`, `set()` y `dict()` crean una estructura a partir de otra.

## 6.1 Convertir a lista

````python
print(list((10, 20, 30)))                       # desde una tupla
print(sorted({20, 40, 60, 80}))                 # desde un conjunto (sorted ya da una lista ordenada)

paises = {"España": "Madrid", "Francia": "París", "Italia": "Roma"}
print(list(paises))                             # solo las claves
print(list(paises.values()))                    # los valores
print(list(paises.items()))                     # los pares

print(list(range(1, 10, 2)))                    # [1, 3, 5, 7, 9]
print(list("Hola Mundo"))                       # cada carácter, incluido el espacio
````

> Para separar una cadena en **palabras** no se usa `list()`, sino `split()`: `"Hola Mundo".split()` da `['Hola', 'Mundo']`.

## 6.2 Convertir a tupla

````python
print(tuple([10, 20, 30]))
print(tuple(sorted({20, 40, 60})))
print(tuple({"España": "Madrid", "Francia": "París"}))    # las claves
````

## 6.3 Convertir a conjunto

Al pasar a conjunto se pierden los duplicados y el orden.

````python
print(sorted(set([10, 20, 30, 20, 10])))        # [10, 20, 30]
print(sorted(set((20, 40, 60, 40))))            # [20, 40, 60]
print(sorted(set({"España": "Madrid", "Francia": "París"})))
````

## 6.4 Convertir a diccionario

`dict()` necesita una colección de **pares**: cada elemento debe tener exactamente dos partes, la clave y el valor.

````python
pares = [("España", "Madrid"), ("Francia", "París"), ("Italia", "Roma")]
print(dict(pares))

paises = ["España", "Francia", "Italia"]
capitales = ["Madrid", "París", "Roma"]
print(dict(zip(paises, capitales)))             # zip empareja las dos listas

try:
    dict([("a", 1, 2)])                         # el elemento tiene 3 partes, no 2
except ValueError as e:
    print("ValueError:", e)
````

# 7. Copias y referencias

## 7.1 La asignación no copia

En Python, una variable es una **etiqueta** que apunta a un objeto. `b = a` no crea una lista nueva: pega una **segunda etiqueta** en la misma lista. Si se modifica a través de `b`, el cambio se ve también en `a`.

```
a ──┐
    ├──►  [1, 2, 3]
b ──┘
```

Esto pasa con todo lo **mutable** (listas, conjuntos, diccionarios). Con números, cadenas y tuplas no hay problema, porque no se pueden modificar.

````python
a = [1, 2, 3]
b = a                 # NO es una copia
b.append(4)
print(a)              # [1, 2, 3, 4]  ¡a también ha cambiado!
print(a is b)         # True: son el mismo objeto
````

## 7.2 Copia superficial

Para tener una lista independiente hay que **copiarla**. Las tres formas siguientes son equivalentes. Los diccionarios y conjuntos también tienen `.copy()`.

````python
a = [1, 2, 3]
b = a.copy()          # o list(a), o a[:]
b.append(4)
print(a)              # [1, 2, 3]
print(b)              # [1, 2, 3, 4]
print(a is b)         # False
````

## 7.3 Copia profunda

`copy()` es una copia **superficial**: crea una lista nueva, pero si dentro hay otras listas (o diccionarios), copia las **referencias** a ellas, no su contenido. Para estructuras anidadas se usa `copy.deepcopy()`, que lo copia todo.

````python
import copy

original = [[1, 2], [3, 4]]
superficial = original.copy()
profunda = copy.deepcopy(original)

original[0].append(99)
print(superficial)    # [[1, 2, 99], [3, 4]]  comparte las sublistas
print(profunda)       # [[1, 2], [3, 4]]      totalmente independiente
````

> **Lo mismo pasa al pasar una lista a una función**: la función recibe la misma lista, no una copia, y si la modifica, el cambio se ve fuera. En pandas verás la misma idea con `df.copy()`.

# 9. Errores frecuentes

| Lo que ves | Qué pasa | Solución |
|---|---|---|
| `IndexError: list index out of range` | Posición que no existe; el último índice es `len(lista) - 1` | Revisa el índice, o usa `lista[-1]` para el último |
| `ValueError: ... not in list` | `remove(x)` o `index(x)` con un valor que no está | Comprueba antes con `x in lista` |
| La variable vale `None` | Se guardó el resultado de `sort()`, `append()`, `reverse()`… que modifican la lista y devuelven `None` | `lista.sort()` sin asignar, o `nueva = sorted(lista)` |
| `[1, 2, [3, 4]]` en lugar de `[1, 2, 3, 4]` | Se usó `append` con una lista | `extend` |
| `TypeError: 'tuple' object does not support item assignment` | Las tuplas no se modifican | Crea una tupla nueva, o usa una lista |
| `(5)` no es una tupla | Falta la coma | `(5,)` |
| `{}` no es un conjunto vacío | `{}` es un diccionario vacío | `set()` |
| `TypeError: 'set' object is not subscriptable` | Los conjuntos no tienen posiciones | `sorted(conjunto)[0]`, o usa una lista |
| `TypeError: unhashable type: 'list'` | Una lista como clave de diccionario o elemento de conjunto | Usa una tupla |
| `KeyError` | La clave no está en el diccionario | `dic.get(clave)` o comprueba con `clave in dic` |
| `"Roma" in capitales` da `False` | `in` busca en las claves | `"Roma" in capitales.values()` |
| Al cambiar una lista cambia otra | `b = a` no copia | `b = a.copy()`, o `copy.deepcopy(a)` si es anidada |
